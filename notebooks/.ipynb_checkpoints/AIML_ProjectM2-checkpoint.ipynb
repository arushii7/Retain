# Stage 3: Data Preprocessing

The dataset is preprocessed to make it suitable for machine learning. This includes removing non-predictive identifiers, separating the target variable, splitting the data into training and testing sets, handling missing values, encoding categorical variables, and scaling numerical features.

```python
from sklearn.model_selection import train_test_split

# Remove CustomerID because it is only an identifier
X = df.drop(columns=['Churn', 'CustomerID'])
y = df['Churn']

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'str']
).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine both
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

# Fit preprocessing ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply same transformations to test data
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

# Stage 4: Feature Selection

L1-regularized Logistic Regression is used for feature selection. The L1 penalty encourages the coefficients of less important features to become zero, allowing features with stronger predictive contributions to be identified and reducing unnecessary model complexity.

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of features after preprocessing:", len(feature_names))
```

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

lasso_model = LogisticRegression(
    solver='liblinear',
    C=0.1,
    l1_ratio=1.0,
    random_state=42,
    max_iter=1000
)

lasso_model.fit(X_train_processed, y_train)
```

```python
coefficients = lasso_model.coef_[0]

feature_importance = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients,
    'Absolute_Coefficient': np.abs(coefficients)
})

feature_importance = feature_importance.sort_values(
    by='Absolute_Coefficient',
    ascending=False
)

feature_importance.head(20)
```

```python
selected_mask = coefficients != 0

selected_features = feature_names[selected_mask]

print("Total features after preprocessing:", len(feature_names))
print("Selected features:", len(selected_features))

print("\nSelected features:")
for feature in selected_features:
    print(feature)
```

```python
top_features = feature_importance.head(15).sort_values(
    by='Coefficient'
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features['Feature'],
    top_features['Coefficient']
)

plt.xlabel('L1 Logistic Regression Coefficient')
plt.ylabel('Feature')
plt.title('Top Features Selected by L1 Regularization')

plt.tight_layout()
plt.show()
```

```python
X_train_selected = X_train_processed[:, selected_mask]
X_test_selected = X_test_processed[:, selected_mask]

print("Final training data shape:", X_train_selected.shape)
print("Final testing data shape:", X_test_selected.shape)
```

# Stage 5: Baseline Model - Logistic Regression

A Logistic Regression model is implemented as the baseline classification model for predicting customer churn. The model is trained using the selected features obtained after preprocessing and L1-based feature selection.

```python
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression baseline model
logistic_model = LogisticRegression(
    random_state=42
)

# Train the model
logistic_model.fit(X_train_selected, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Predictions on test data
y_pred = logistic_model.predict(X_test_selected)

# Predicted probabilities for the churn class
y_prob = logistic_model.predict_proba(X_test_selected)[:, 1]

print("Predictions generated successfully.")
print("Number of test predictions:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("===== LOGISTIC REGRESSION BASELINE RESULTS =====")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")
```

```python
from sklearn.metrics import classification_report

print("===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_test,
    y_pred,
    target_names=['No Churn', 'Churn']
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['No Churn', 'Churn']
)

disp.plot()
plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

```python
baseline_results = pd.DataFrame({
    'Metric': [
        'Accuracy',
        'Precision',
        'Recall',
        'F1-Score'
    ],
    'Score': [
        accuracy,
        precision,
        recall,
        f1
    ]
})

baseline_results['Score'] = baseline_results['Score'].round(4)

display(baseline_results)
```

# Stage 6: Decision Tree

A Decision Tree classifier is implemented to capture non-linear relationships between customer features and churn. Unlike Logistic Regression, which models a linear decision boundary, a Decision Tree makes predictions using a sequence of feature-based decisions.

```python
from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

decision_tree.fit(X_train_selected, y_train)

y_pred_dt = decision_tree.predict(X_test_selected)

print("Decision Tree model trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

dt_accuracy = accuracy_score(y_test, y_pred_dt)
dt_precision = precision_score(y_test, y_pred_dt)
dt_recall = recall_score(y_test, y_pred_dt)
dt_f1 = f1_score(y_test, y_pred_dt)

print("===== DECISION TREE RESULTS =====")
print(f"Accuracy  : {dt_accuracy:.4f}")
print(f"Precision : {dt_precision:.4f}")
print(f"Recall    : {dt_recall:.4f}")
print(f"F1-Score  : {dt_f1:.4f}")

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_test,
    y_pred_dt,
    target_names=['No Churn', 'Churn']
))
```

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_dt,
    display_labels=['No Churn', 'Churn']
)

disp.plot()
plt.title('Decision Tree - Confusion Matrix')
plt.show()
```

A Random Forest classifier is implemented as an ensemble learning model. It combines predictions from multiple Decision Trees to improve robustness and reduce the risk of relying on a single tree.

# Stage 8: Gradient Boosting

Gradient Boosting is implemented as an ensemble learning model. It builds decision trees sequentially, with each new tree attempting to correct the errors made by the previous trees. This allows the model to capture non-linear relationships and complex patterns in customer churn data.

```python
from sklearn.ensemble import GradientBoostingClassifier

gradient_boosting = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gradient_boosting.fit(X_train_selected, y_train)

y_pred_gb = gradient_boosting.predict(X_test_selected)

print("Gradient Boosting model trained successfully.")
```

```python
gb_accuracy = accuracy_score(y_test, y_pred_gb)
gb_precision = precision_score(y_test, y_pred_gb)
gb_recall = recall_score(y_test, y_pred_gb)
gb_f1 = f1_score(y_test, y_pred_gb)

print("===== GRADIENT BOOSTING RESULTS =====")
print(f"Accuracy  : {gb_accuracy:.4f}")
print(f"Precision : {gb_precision:.4f}")
print(f"Recall    : {gb_recall:.4f}")
print(f"F1-Score  : {gb_f1:.4f}")

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_test,
    y_pred_gb,
    target_names=['No Churn', 'Churn']
))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.feature_selection import SelectFromModel
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
import pandas as pd
import numpy as np
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

final_model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),

    ('feature_selection', SelectFromModel(
        estimator=LogisticRegression(
            solver='liblinear',
            C=0.1,
            l1_ratio=1.0,
            max_iter=1000,
            random_state=42
        ),
        threshold=1e-8
    )),

    ('classifier', RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ))
])

print("Final Random Forest pipeline created successfully.")
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    final_model_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring={
        'accuracy': 'accuracy',
        'precision': 'precision',
        'recall': 'recall',
        'f1': 'f1'
    },
    return_train_score=False
)

print("===== 5-FOLD CROSS VALIDATION RESULTS =====")

for metric in ['accuracy', 'precision', 'recall', 'f1']:
    scores = cv_results[f'test_{metric}']

    print(f"{metric.capitalize():10s}: "
          f"{scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
cv_summary = pd.DataFrame({
    'Fold': [1, 2, 3, 4, 5],
    'Accuracy': cv_results['test_accuracy'],
    'Precision': cv_results['test_precision'],
    'Recall': cv_results['test_recall'],
    'F1-Score': cv_results['test_f1']
})

display(cv_summary.round(4))
```

```python
final_model_pipeline.fit(X_train, y_train)

print("Final Random Forest pipeline trained successfully.")
```

```python
final_predictions = final_model_pipeline.predict(X_test)

final_probabilities = final_model_pipeline.predict_proba(X_test)[:, 1]

print("Predictions generated successfully.")
print("Total test predictions:", len(final_predictions))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

final_accuracy = accuracy_score(y_test, final_predictions)
final_precision = precision_score(y_test, final_predictions)
final_recall = recall_score(y_test, final_predictions)
final_f1 = f1_score(y_test, final_predictions)

print("===== FINAL RANDOM FOREST RESULTS =====")

print(f"Accuracy  : {final_accuracy:.4f}")
print(f"Precision : {final_precision:.4f}")
print(f"Recall    : {final_recall:.4f}")
print(f"F1-Score  : {final_f1:.4f}")

print("\n===== CLASSIFICATION REPORT =====")

print(classification_report(
    y_test,
    final_predictions,
    target_names=['No Churn', 'Churn']
))
```

Customer Segmentation using K-Means Clustering

```python
from sklearn.base import clone
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Create a fresh copy of the preprocessing pipeline
cluster_preprocessor = clone(preprocessor)

# Transform all customer features without using Churn
X_cluster_processed = cluster_preprocessor.fit_transform(X)

print("Clustering dataset prepared successfully.")
print("Processed shape:", X_cluster_processed.shape)
```

```python
cluster_scores = []

for k in range(2, 9):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_cluster_processed)

    inertia = kmeans.inertia_

    silhouette = silhouette_score(
        X_cluster_processed,
        labels,
        sample_size=2000,
        random_state=42
    )

    cluster_scores.append({
        'K': k,
        'Inertia': inertia,
        'Silhouette Score': silhouette
    })

cluster_results = pd.DataFrame(cluster_scores)

display(cluster_results.round(4))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    cluster_results['K'],
    cluster_results['Inertia'],
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(range(2, 9))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    cluster_results['K'],
    cluster_results['Silhouette Score'],
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different K Values')
plt.xticks(range(2, 9))
plt.grid(True)
plt.show()
```

```python
best_k = int(
    cluster_results.loc[
        cluster_results['Silhouette Score'].idxmax(),
        'K'
    ]
)

print("Selected number of clusters:", best_k)

kmeans_model = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans_model.fit_predict(
    X_cluster_processed
)

print("K-Means model trained successfully.")
```

```python
segmentation_results = X.copy()

segmentation_results['Cluster'] = cluster_labels

# Add churn labels only for post-clustering analysis
segmentation_results['Churn'] = df.loc[X.index, 'Churn']

display(segmentation_results.head())
```

```python
cluster_distribution = segmentation_results['Cluster'].value_counts().sort_index()

print("===== CUSTOMER DISTRIBUTION =====")

display(
    cluster_distribution.rename('Customer Count').to_frame()
)
```

```python
cluster_profile = segmentation_results.groupby('Cluster')[
    numeric_features
].mean().round(2)

display(cluster_profile)
```

```python
cluster_churn = segmentation_results.groupby('Cluster').agg(
    Customer_Count=('Churn', 'count'),
    Churned_Customers=('Churn', 'sum'),
    Churn_Rate=('Churn', 'mean')
)

cluster_churn['Churn_Rate'] = (
    cluster_churn['Churn_Rate'] * 100
).round(2)

display(cluster_churn)
```

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

cluster_2d = svd.fit_transform(X_cluster_processed)

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    cluster_2d[:, 0],
    cluster_2d[:, 1],
    c=cluster_labels,
    cmap='viridis',
    alpha=0.6
)

plt.xlabel('Component 1')
plt.ylabel('Component 2')
plt.title('Customer Segmentation Using K-Means')

plt.colorbar(scatter, label='Cluster')
plt.show()
```

```python
all_churn_probabilities = final_model_pipeline.predict_proba(X)[:, 1]

customer_analysis = X.copy()

customer_analysis['CustomerID'] = df.loc[X.index, 'CustomerID']

customer_analysis['Churn_Probability'] = all_churn_probabilities

customer_analysis['Predicted_Churn'] = (
    all_churn_probabilities >= 0.5
).astype(int)

customer_analysis['Cluster'] = cluster_labels

print("Customer churn analysis generated successfully.")

display(customer_analysis.head())
```

```python
def assign_risk(probability):

    if probability < 0.30:
        return 'Low'

    elif probability < 0.60:
        return 'Medium'

    else:
        return 'High'


customer_analysis['Risk_Category'] = (
    customer_analysis['Churn_Probability']
    .apply(assign_risk)
)

display(
    customer_analysis[
        ['CustomerID', 'Churn_Probability',
         'Risk_Category', 'Cluster']
    ].head(10)
)
```

```python
def generate_recommendation(row):

    recommendations = []

    if row['Churn_Probability'] < 0.30:
        return "Maintain engagement through loyalty rewards."

    if row['Complain'] == 1:
        recommendations.append(
            "Prioritise customer support and complaint resolution."
        )

    if row['SatisfactionScore'] <= 2:
        recommendations.append(
            "Offer personalised service recovery."
        )

    if row['DaySinceLastOrder'] >= 7:
        recommendations.append(
            "Launch a personalised re-engagement campaign."
        )

    if row['OrderCount'] <= 2:
        recommendations.append(
            "Offer a first-time or repeat-purchase incentive."
        )

    if row['Tenure'] <= 2:
        recommendations.append(
            "Introduce an early customer onboarding campaign."
        )

    if row['Cluster'] == 1:
        recommendations.append(
            "Consider a loyalty reward for this established customer."
        )

    if not recommendations:
        recommendations.append(
            "Send personalised product recommendations and monitor engagement."
        )

    return " ".join(recommendations)
```

```python
customer_analysis['Retention_Recommendation'] = (
    customer_analysis.apply(
        generate_recommendation,
        axis=1
    )
)

display(
    customer_analysis[
        [
            'CustomerID',
            'Churn_Probability',
            'Risk_Category',
            'Cluster',
            'Retention_Recommendation'
        ]
    ].head(15)
)
```
